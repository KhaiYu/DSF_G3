## 2. Data Source & Understanding

In this phase of the analysis, we return to the Pandas library for data restructuring. 
This process will include indexing DataFrames, sorting variables, and performing grouping operations in order to prepare the data for further analysis. For the start of our technical pipeline, the following packages are imported.

```python
import numpy as np
import pandas as pd
```

Beyond simple exercises, we can now employ Pandas for a much larger and more complex data set: the E-commerce Shipping Data. This is a particularly useful data set for logistics purposes and contains important metadata, including shipping mode, cost of the goods being transported, and the discounts applied, along with the delivery status. In order to kick off our technical process, we will use the `read_csv` command.

```python
train = pd.read_csv("Train.csv")
train
```

**2.1 Description**

**Features:** 12 columns including `ID`, `Warehouse_block`, `Mode_of_Shipment`,` Customer_care_calls`, `Customer_rating`, `Cost_of_the_Product`, `Prior_purchases`,`Product_importance`,` Gender`, `Discount_offered`, `Weight_in_gms`,` Reached.on.Time_Y.N`.

**Size:** 10999 observations

**Units:** Weight in grams (gms), Cost in USD.

**2.2 Limitation**

**Coverage**

Although the main focus of the data set is on global B2C e-commerce deliveries, it does not provide any detailed geographic information, for example, the cities where the packages have been sent or the weather during transportation.

**Missingness**

A typical way to test for the presence of null values using `df.isnull().sum()` suggests there are no null values; however, there can be hidden missing values. An example would be a value of '0' in the column Discount_offered, which means no discount was offered.

**Sampling Bias**

The data may exhibit survivorship bias. It only includes orders that were successfully processed and reached a final status. Orders that were lost in transit, cancelled by the customer, or returned before reaching the destination are likely excluded from this specific dataset.

**Known Caveats**

The Customer_rating column is highly subjective. A low rating might reflect a customer's dissatisfaction with the produ
ct quality itself rather than the logistics performance, potentially creating "noise" when analyzing shipping efficiency.

## 4. Data Cleaning & Preprocessing
- Handle missing values (strategy and justification)
- Fix types (dates, categoricals, numerics)
- Remove/flag outliers

To conduct an accurate diagnosis of the completeness of the data set, a combination of the 
`.isnull()` and `.sum()` methods needs to be used. As while the former finds null cells in the data set,
the latter method will help to find out the number of these cases for each particular column.
The importance of such quantitative diagnostics stems from the ability to understand the extent of the issue and deal with it by either deleting data or using imputing methods.


```python
print("Checking of Missing values")
train.isnull().sum()
```

Hidden missingness occurs when data is present but logically incorrect. 
For example,`Weight_in_gms` a shipping weight of 0 grams or `Cost_of_the_product` of $0 is impossible in real logistics.
We checked our data for these 'hidden' errors to make sure our analysis is based on real, accurate information

```python
hidden_missing_weight = (train['Weight_in_gms'] == 0).sum()
hidden_missing_cost = (train['Cost_of_the_Product'] == 0).sum()

print(f"Number of anomalies in Weight (Weight = 0): {hidden_missing_weight}")
print(f"Number of anomalies in Product Cost (Cost = 0): {hidden_missing_cost}")
```

**Is the data entered by a human?**

Whereas ID numbers and monetary figures are system-generated, categorical values are usually created manually by people. We need to assess whether the entries in those fields are accurate enough by looking at some categorical fields. For instance, we can look at the unique values in fields such as `Mode_of_Shipment` and `Product_importance` to see whether they are consistently using vocabulary.

```python
cat_cols = ['Mode_of_Shipment', 'Product_importance']
print("Cheking unique value for Mode_of_Shipment and Product_importance")

for col in cat_cols:
    print(f"{col}: {train[col].unique()}")
```
